```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from PIL import Image
import torch

print("Loading Moondream2 (The New Brain)... This may take a minute.")

model_id = "vikhyatk/moondream2"

# We load the model in float16 to make it run incredibly fast on the T4 GPU
# If you encounter 'AssertionError: Torch not compiled with CUDA enabled',
# ensure your Colab runtime is set to a GPU type (e.g., T4 GPU).
# Otherwise, the model will run on CPU.
device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.float16 if torch.cuda.is_available() else torch.float32

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    trust_remote_code=True,
    torch_dtype=dtype
).to(device)

tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True)

print("New Brain Loaded Successfully!")

def analyze_visual_context_v2(image_path, user_question):
    """Processes the image and text using Moondream2."""
    try:
        # 1. Open the image
        raw_image = Image.open(image_path)

        # 2. The AI 'looks' at the image and encodes it into memory
        enc_image = model.encode_image(raw_image)

        # 3. The AI answers the question based on that visual memory
        answer = model.answer_question(enc_image, user_question, tokenizer)
        return answer

    except Exception as e:
        return f"Error analyzing visual data: {e}"

# --- TEST THE NEW BRAIN ---
test_image = "current_capture.jpg"
user_query = "What specific object is the person holding in their hand?"

result = analyze_visual_context_v2(test_image, user_query)

print("-" * 30)
print(f"QUESTION: {user_query}")
print(f"PERSONA'S BRAIN SAYS: {result}")
print("-" * 30)
```
